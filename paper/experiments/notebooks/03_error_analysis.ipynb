# Error analysis
The notebook joins claims and predictions without redefining metrics or model logic.

```python
from pathlib import Path
import json, sys
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
from data.io import read_jsonl
claims = {row['claim_id']: row for row in read_jsonl(root / 'outputs' / 'smoke' / 'claims.jsonl')}
predictions = read_jsonl(root / 'outputs' / 'smoke' / 'predictions.jsonl')
errors = [{'claim_id': p['claim_id'], 'truth': claims[p['claim_id']]['normalized_label'], 'prediction': p['predicted_label']} for p in predictions if claims[p['claim_id']]['normalized_label'] != p['predicted_label']]
{'error_count': len(errors), 'errors': errors}
```
